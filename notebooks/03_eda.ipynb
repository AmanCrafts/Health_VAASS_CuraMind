```python

# ==========================================
# 03_eda.ipynb
# ==========================================

import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------
# Load Dataset
# ------------------------------------------

DATA_PATH = "../data/processed/cleaned_data.csv"

df = pd.read_csv(DATA_PATH)

print("Dataset Loaded Successfully")

# ------------------------------------------
# Basic Information
# ------------------------------------------

print("\nDataset Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nDataset Info:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())
```

```python

# ==========================================
# MISSING VALUES ANALYSIS
# ==========================================

print("\nMissing Values:")
print(df.isnull().sum())
```

```python

# ==========================================
# AGE DISTRIBUTION
# ==========================================

plt.figure(figsize=(8,5))

plt.hist(df["age"], bins=20)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python

# ==========================================
# GENDER DISTRIBUTION
# ==========================================

gender_counts = df["gender"].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    gender_counts,
    labels=gender_counts.index,
    autopct="%1.1f%%"
)

plt.title("Gender Distribution")

plt.show()
```

```python

# ==========================================
# CITY DISTRIBUTION
# ==========================================

city_counts = df["city"].value_counts()

plt.figure(figsize=(10,5))

plt.bar(city_counts.index, city_counts.values)

plt.title("Patients by City")
plt.xlabel("City")
plt.ylabel("Count")

plt.xticks(rotation=45)

plt.show()
```

```python

# ==========================================
# BMI DISTRIBUTION
# ==========================================

plt.figure(figsize=(8,5))

plt.hist(df["bmi"], bins=20)

plt.title("BMI Distribution")
plt.xlabel("BMI")
plt.ylabel("Frequency")

plt.show()
```

```python

# ==========================================
# HEART RATE DISTRIBUTION
# ==========================================

plt.figure(figsize=(8,5))

plt.hist(df["heart_rate"], bins=20)

plt.title("Heart Rate Distribution")
plt.xlabel("Heart Rate")
plt.ylabel("Frequency")

plt.show()
```

```python

# ==========================================
# CHOLESTEROL DISTRIBUTION
# ==========================================

plt.figure(figsize=(8,5))

plt.hist(df["cholesterol_level"], bins=20)

plt.title("Cholesterol Level Distribution")
plt.xlabel("Cholesterol")
plt.ylabel("Frequency")

plt.show()
```

```python

# ==========================================
# DIABETIC STATUS
# ==========================================

diabetic_counts = df["diabetic"].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    diabetic_counts,
    labels=["Non-Diabetic", "Diabetic"],
    autopct="%1.1f%%"
)

plt.title("Diabetic Distribution")

plt.show()
```

```python

# ==========================================
# SMOKER STATUS
# ==========================================

smoker_counts = df["smoker"].value_counts()

plt.figure(figsize=(8,5))

plt.bar(smoker_counts.index, smoker_counts.values)

plt.title("Smoking Status")
plt.xlabel("Smoking Category")
plt.ylabel("Count")

plt.xticks(rotation=20)

plt.show()
```

```python

# ==========================================
# HAS DISEASE DISTRIBUTION
# ==========================================

disease_counts = df["has_disease"].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    disease_counts,
    labels=["No Disease", "Has Disease"],
    autopct="%1.1f%%"
)

plt.title("Disease Distribution")

plt.show()
```

```python

# ==========================================
# BLOOD PRESSURE ANALYSIS
# ==========================================

plt.figure(figsize=(8,5))

plt.hist(df["systolic_bp"], bins=20)

plt.title("Systolic Blood Pressure Distribution")
plt.xlabel("Systolic BP")
plt.ylabel("Frequency")

plt.show()

plt.figure(figsize=(8,5))

plt.hist(df["diastolic_bp"], bins=20)

plt.title("Diastolic Blood Pressure Distribution")
plt.xlabel("Diastolic BP")
plt.ylabel("Frequency")

plt.show()
```

```python

# ==========================================
# CORRELATION ANALYSIS
# ==========================================

numeric_cols = [
    "age",
    "bmi",
    "heart_rate",
    "cholesterol_level",
    "systolic_bp",
    "diastolic_bp",
    "has_disease"
]

correlation_matrix = df[numeric_cols].corr()

print("\nCorrelation Matrix:")
print(correlation_matrix)
```

```python

# ==========================================
# MANUAL HEATMAP USING MATPLOTLIB
# ==========================================

plt.figure(figsize=(8,6))

plt.imshow(correlation_matrix)

plt.colorbar()

plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=45
)

plt.yticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)

plt.title("Correlation Heatmap")

plt.show()
```

```python

# ==========================================
# AGE VS CHOLESTEROL
# ==========================================

plt.figure(figsize=(8,5))

plt.scatter(
    df["age"],
    df["cholesterol_level"]
)

plt.title("Age vs Cholesterol Level")
plt.xlabel("Age")
plt.ylabel("Cholesterol")

plt.show()
```

```python

# ==========================================
# BMI VS HEART RATE
# ==========================================

plt.figure(figsize=(8,5))

plt.scatter(
    df["bmi"],
    df["heart_rate"]
)

plt.title("BMI vs Heart Rate")
plt.xlabel("BMI")
plt.ylabel("Heart Rate")

plt.show()
```

```python

# ==========================================
# TOP DIAGNOSIS CODES
# ==========================================

diagnosis_counts = df["diagnosis_code"].value_counts().head(10)

plt.figure(figsize=(10,5))

plt.bar(
    diagnosis_counts.index,
    diagnosis_counts.values
)

plt.title("Top Diagnosis Codes")
plt.xlabel("Diagnosis Code")
plt.ylabel("Count")

plt.show()
```

```python

# ==========================================
# HIGH RISK PATIENTS
# ==========================================

high_risk = df[
    (df["cholesterol_level"] > 240) |
    (df["bmi"] > 30) |
    (df["heart_rate"] > 100)
]

print("\nHigh Risk Patients Count:")
print(high_risk.shape[0])
```

```python

# ==========================================
# SUMMARY INSIGHTS
# ==========================================

print("\nEDA SUMMARY")

print(f"Total Patients: {len(df)}")

print(f"Average Age: {df['age'].mean():.2f}")

print(f"Average BMI: {df['bmi'].mean():.2f}")

print(f"Average Heart Rate: {df['heart_rate'].mean():.2f}")

print(f"Average Cholesterol: {df['cholesterol_level'].mean():.2f}")

print(f"Disease Positive Patients: {df['has_disease'].sum()}")

print(f"Diabetic Patients: {df['diabetic'].sum()}")

print("\nEDA Completed Successfully")
```
